# Performance analysis

Load real published measurements. Quality is a small diagnostic probe; no synthetic benchmark results are generated here.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
run = root / "results" / "published"
df = pd.read_csv(run / "benchmark_results.csv")
print(df.groupby("status").size())
measured = df[df.status.eq("ok")].copy()
measured.head()
```

```python
keys = ["model_id", "precision", "batch_size", "sequence_length"]
measured.groupby(keys)[["latency_mean_ms", "throughput_samples_s", "sampled_inference_peak_rss_mib", "serialized_state_mib"]].agg(["mean", "std"])
```

```python
import plotly.express as px
px.box(measured, x="model_id", y="latency_mean_ms", color="precision", points="all", title="Independent process trials: mean forward latency")
```
